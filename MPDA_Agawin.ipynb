# 1. WHAT DATE HAS THE HIGHEST AMOUNT OF ORDER?

```python
# Count orders by date and find the date with highest orders
orders_by_date = amazondata.groupby('OrderDate')['OrderID'].count().reset_index()
orders_by_date.columns = ['OrderDate', 'OrderCount']
orders_by_date = orders_by_date.sort_values('OrderCount', ascending=False)

# Plot
sns.lineplot(data=orders_by_date, x='OrderDate', y='OrderCount', color='green', marker='o')
plt.title('Date With the Highest Number of Orders')
plt.xlabel('Order Date')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Display the date with highest orders
print(f"Date with highest orders: {orders_by_date.iloc[0]['OrderDate']} with {orders_by_date.iloc[0]['OrderCount']} orders")
```

**INSIGHTS** : The visualization indicates that the highest number of orders occurred on April 28, 2022, with a total of 85 orders. However, due to the high density of data points, the order dates in the chart appear stacked and are not clearly distinguishable. To address this limitation, a direct data extraction and aggregation were performed, which confirmed that April 28, 2022 recorded the peak order volume. This highlights the importance of combining visual analysis with numerical validation to ensure accuracy when interpreting densely populated time series data.

# 3. What month/year generates the highest total revenue?

```python
# Convert OrderDate to datetime and aggregate revenue by month-year
amazondata['OrderDate'] = pd.to_datetime(amazondata['OrderDate'])
amazondata['MonthYear'] = amazondata['OrderDate'].dt.to_period('M').dt.to_timestamp()
revenue_by_month = amazondata.groupby('MonthYear')['TotalAmount'].sum().reset_index().sort_values('MonthYear')

# Identify top month-year
top = revenue_by_month.loc[revenue_by_month['TotalAmount'].idxmax()]

# Plot revenue by month-year and highlight the highest
plt.figure(figsize=(14,6))
colors = ['orange' if dt == top['MonthYear'] else 'skyblue' for dt in revenue_by_month['MonthYear']]
plt.bar(revenue_by_month['MonthYear'].dt.strftime('%Y-%m'), revenue_by_month['TotalAmount'], color=colors)
plt.xticks(rotation=45)
plt.ylabel('Total Revenue (TotalAmount)')
plt.title('Monthly Revenue by Month-Year')
plt.tight_layout()
plt.show()

print(f"Month/year with highest total revenue: {top['MonthYear'].strftime('%Y-%m')} with ${top['TotalAmount']:.2f}")
```

**INSIGHTS** : The month of May 2024 (2024-05) recorded the highest total revenue, amounting to $1,642,609.94. This indicates a strong surge in customer purchasing activity during this period, which may be attributed to factors such as seasonal demand, promotional campaigns, or increased product availability. The peak in revenue suggests that May is a critical month for sales performance, and Amazon could benefit from reinforcing marketing efforts, inventory readiness, and logistics planning during this period to maximize future revenue potential.

# 4. Are there seasonal trends in order volume?

```python
# Seasonal trends in order volume: heatmap (month vs year) and average monthly orders across years
amazondata['OrderYear'] = amazondata['OrderDate'].dt.year
amazondata['OrderMonth'] = amazondata['OrderDate'].dt.month

monthly_counts = amazondata.groupby(['OrderYear', 'OrderMonth'])['OrderID'].count().reset_index(name='OrderCount')
pivot = monthly_counts.pivot(index='OrderMonth', columns='OrderYear', values='OrderCount').reindex(range(1,13))

# Heatmap + Avg monthly orders
plt.figure(figsize=(16,6))

# Heatmap
plt.subplot(1,2,1)
sns.heatmap(pivot, annot=True, fmt='.0f', cmap='YlGnBu', cbar_kws={'label':'Order Count'}, linewidths=.5)
plt.title('Monthly Order Volume (Month vs Year)')
plt.xlabel('Year')
plt.ylabel('Month')
month_labels = pd.to_datetime(pivot.index.astype(str), format='%m').month_name()
plt.yticks(np.arange(pivot.shape[0]) + 0.5, labels=month_labels, rotation=0)

# Average monthly orders with variability
plt.subplot(1,2,2)
avg_orders = monthly_counts.groupby('OrderMonth')['OrderCount'].mean().reindex(range(1,13))
std_orders = monthly_counts.groupby('OrderMonth')['OrderCount'].std().reindex(range(1,13))
months = range(1,13)
plt.plot(months, avg_orders, marker='o', color='tomato')
plt.fill_between(months, avg_orders - std_orders, avg_orders + std_orders, color='tomato', alpha=0.2)
plt.xticks(months, pd.to_datetime(pd.Series(months).astype(str), format='%m').dt.month_name().str[:3])
plt.title('Average Monthly Orders Across Years')
plt.xlabel('Month')
plt.ylabel('Average Order Count')

# Annotate peak month
peak_month = int(avg_orders.idxmax())
peak_val = avg_orders.max()
plt.annotate(f'{pd.to_datetime(str(peak_month), format="%m").month_name()} ({int(peak_val)})',
             xy=(peak_month, peak_val), xytext=(peak_month, peak_val + std_orders.max()*1.1),
             ha='center', arrowprops=dict(arrowstyle='->'))

plt.tight_layout()
plt.show()
```

**INSIGHTS** : Order volume demonstrates a clear seasonal trend, with August recording the highest average monthly orders (1,733) across multiple years. This consistent peak suggests that customer demand intensifies during this period, potentially driven by seasonal shopping behavior, back-to-school purchases, or promotional events. Recognizing August as a high-demand month allows for better forecasting, inventory optimization, and targeted marketing strategies to capitalize on increased customer activity.


# 5. How does total sales vary over time?

```python
# Visualize how total sales vary over time (monthly totals with rolling averages)
rev = revenue_by_month.copy()
rev['MonthYear'] = pd.to_datetime(rev['MonthYear'])
rev = rev.sort_values('MonthYear').set_index('MonthYear')

# Rolling averages
rev['rolling_3m'] = rev['TotalAmount'].rolling(3, min_periods=1).mean()
rev['rolling_12m'] = rev['TotalAmount'].rolling(12, min_periods=1).mean()

# Plot
import matplotlib.ticker as mtick
plt.figure(figsize=(14,6))
plt.plot(rev.index, rev['TotalAmount'], label='Monthly Total Revenue', marker='o', color='tab:blue')
plt.plot(rev.index, rev['rolling_3m'], label='3-month Rolling Mean', linestyle='--', color='orange')
plt.plot(rev.index, rev['rolling_12m'], label='12-month Rolling Mean', linestyle=':', color='green')

# Highlight peak month
peak_date = rev['TotalAmount'].idxmax()
peak_val = rev['TotalAmount'].max()
plt.scatter([peak_date], [peak_val], color='red', zorder=5, s=80)
plt.annotate(f'Peak: {peak_date.strftime("%Y-%m")}\n${peak_val:,.0f}',
             xy=(peak_date, peak_val), xytext=(0, 40), textcoords='offset points',
             ha='center', arrowprops=dict(arrowstyle='->', color='red'))

# Formatting
plt.title('Total Sales Over Time (Monthly Revenue)')
plt.xlabel('Month-Year')
plt.ylabel('Total Revenue ($)')
plt.xticks(rotation=45)
plt.gca().yaxis.set_major_formatter(mtick.StrMethodFormatter('${x:,.0f}'))
plt.grid(alpha=0.25)
plt.legend()
plt.tight_layout()
plt.show()
```

**INSIGHTS** : Total sales exhibit clear variation over time, with a pronounced peak in May 2024, when revenue reached approximately $1,642,610. This significant surge indicates a period of exceptionally strong sales performance, likely influenced by seasonal demand, major promotional activities, or increased customer purchasing behavior. The result highlights May as a strategically important month, suggesting that reinforcing marketing efforts, inventory planning, and operational support during this period could help sustain or even amplify future revenue growth.

# 9. How does shipping cost affect the Total Amount?

```python
# Scatter, regression, and binned analysis: how ShippingCost affects TotalAmount
df = amazondata.copy()
df = df[['ShippingCost', 'TotalAmount']].dropna().reset_index(drop=True)
df['log_Shipping'] = np.log1p(df['ShippingCost'])
df['log_Total'] = np.log1p(df['TotalAmount'])

# correlations
corr_raw = df['ShippingCost'].corr(df['TotalAmount'])
corr_log = df['log_Shipping'].corr(df['log_Total'])

# sample for scatter to improve rendering
sample = df.sample(n=10000, random_state=42) if len(df) > 10000 else df

# set plotting limits to avoid extreme outliers dominating the axes
xlim_raw = df['ShippingCost'].quantile(0.99)
ylim_raw = df['TotalAmount'].quantile(0.99)

plt.figure(figsize=(16,12))

# 1) Raw scatter + regression
plt.subplot(2,2,1)
sns.scatterplot(data=sample, x='ShippingCost', y='TotalAmount', alpha=0.25, s=10)
sns.regplot(data=sample, x='ShippingCost', y='TotalAmount', scatter=False, color='red', line_kws={'linewidth':2})
plt.xlim(0, xlim_raw)
plt.ylim(0, ylim_raw)
plt.title(f'ShippingCost vs TotalAmount (raw)   Corr={corr_raw:.3f}')
plt.xlabel('ShippingCost')
plt.ylabel('TotalAmount ($)')

# 2) Log-transformed scatter + regression
plt.subplot(2,2,2)
sns.scatterplot(data=sample, x='log_Shipping', y='log_Total', alpha=0.25, s=10)
sns.regplot(data=sample, x='log_Shipping', y='log_Total', scatter=False, color='red', line_kws={'linewidth':2})
plt.title(f'log1p(ShippingCost) vs log1p(TotalAmount)   Corr={corr_log:.3f}')
plt.xlabel('log1p(ShippingCost)')
plt.ylabel('log1p(TotalAmount)')

# 3) Boxplot of TotalAmount for ShippingCost quantile bins
plt.subplot(2,2,3)
df['ship_quantile'] = pd.qcut(df['ShippingCost'], 8, duplicates='drop')
sns.boxplot(data=df, x='ship_quantile', y='TotalAmount', palette='coolwarm')
plt.xticks(rotation=45)
plt.title('Distribution of TotalAmount by ShippingCost Quantiles')
plt.xlabel('ShippingCost Quantile Bin')
plt.ylabel('TotalAmount ($)')
plt.ylim(0, ylim_raw)

# 4) Average TotalAmount by ShippingCost bins (line)
plt.subplot(2,2,4)
xmax = df['ShippingCost'].quantile(0.975)
bins = np.linspace(0, xmax, 12)
df['ship_bin'] = pd.cut(df['ShippingCost'], bins=bins, include_lowest=True)
agg = df.groupby('ship_bin')['TotalAmount'].agg(['mean','median','count']).reset_index()
sns.lineplot(data=agg, x=agg.index, y='mean', marker='o', color='tab:blue')
plt.fill_between(agg.index, agg['mean'] - agg['mean'].std(), agg['mean'] + agg['mean'].std(), color='tab:blue', alpha=0.1)
plt.xticks(agg.index, labels=[f'{int(interval.left)}-{int(interval.right)}' for interval in agg['ship_bin']], rotation=45)
plt.title('Average TotalAmount by ShippingCost Bin')
plt.xlabel('ShippingCost Bin ($)')
plt.ylabel('Mean TotalAmount ($)')

plt.tight_layout()
plt.show()

print(f'Correlation (raw): {corr_raw:.3f}, Correlation (log-log): {corr_log:.3f}')
```

**INSIGHTS** : The correlation analysis indicates that shipping cost has a very weak relationship with the Total Amount of an order. The raw correlation of 0.039 suggests almost no linear association, meaning higher shipping costs do not significantly correspond to higher or lower total order values. Even after applying a log–log transformation, the correlation increases only slightly to 0.080, which still reflects a weak positive relationship. This implies that shipping costs have minimal influence on how much customers spend overall, and purchasing decisions are likely driven more by product price, quantity, or discounts rather than shipping fees.

# 12. What percentage of orders are Delivered vs Cancelled?

```python
# Percentage of orders: Delivered vs Cancelled (plus Other)
order_counts = amazondata['OrderStatus'].value_counts()
delivered = order_counts.get('Delivered', 0)
cancelled = order_counts.get('Cancelled', 0)
other = order_counts.sum() - delivered - cancelled

labels = ['Delivered', 'Cancelled', 'Other']
sizes = [delivered, cancelled, other]
colors = ['#2ca02c', '#d62728', 'lightgray']
explode = (0.05, 0.05, 0.0)

plt.figure(figsize=(12,5))

# Pie chart
plt.subplot(1,2,1)
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140, colors=colors, explode=explode, wedgeprops={'edgecolor':'k'})
plt.title('Order Status Distribution (Delivered vs Cancelled)')
plt.axis('equal')

# Horizontal bar chart with counts
plt.subplot(1,2,2)
sns.barplot(x=sizes, y=labels, palette=colors)
for i, v in enumerate(sizes):
    plt.text(v + max(sizes)*0.01, i, f'{v} ({v/sum(sizes)*100:.1f}%)', va='center')
plt.xlabel('Number of Orders')
plt.title('Order Status Counts')

plt.tight_layout()
plt.show()

print(f"Delivered: {delivered} ({delivered/order_counts.sum()*100:.1f}%)")
print(f"Cancelled: {cancelled} ({cancelled/order_counts.sum()*100:.1f}%)")
print(f"Other statuses: {other} ({other/order_counts.sum()*100:.1f}%)")
```

**INSIGHTS** : The majority of orders in the dataset are successfully delivered, accounting for 74.6% (74,628 orders), which reflects strong fulfillment and logistics performance. In contrast, only 3.0% (3,028 orders) of orders are cancelled, indicating a relatively low cancellation rate and generally stable customer purchasing behavior. However, a notable 22.3% of orders fall under other statuses, suggesting the presence of orders that are pending, returned, or in transit. This highlights an opportunity to further investigate and optimize these intermediate order states to improve overall delivery completion and customer satisfaction.

# 13. How does discount percentage affect sales volume?

```python
# Analyze how discount percentage affects sales volume
discount_col = next((c for c in amazondata.columns if 'discount' in c.lower()), None)

if discount_col is None:
    print("No discount column found. Available columns:", amazondata.columns.tolist())
else:
    df_disc = amazondata[[discount_col, 'TotalAmount', 'OrderID']].copy().dropna()
    # Clean/convert discount values (handle strings like '10%' or fractions like 0.1)
    df_disc[discount_col] = df_disc[discount_col].astype(str).str.replace('%', '', regex=False)
    df_disc[discount_col] = pd.to_numeric(df_disc[discount_col], errors='coerce')
    # If values are in [0,1], convert to percent
    if df_disc[discount_col].max() <= 1:
        df_disc[discount_col] = df_disc[discount_col] * 100
    df_disc = df_disc.dropna(subset=[discount_col, 'TotalAmount'])

    # Summary stats / correlation
    corr = df_disc[discount_col].corr(df_disc['TotalAmount'])
    print(f'Correlation (Discount% vs TotalAmount): {corr:.3f}')

    # Scatter with regression (limit to reasonable quantiles to avoid outlier distortion)
    sample = df_disc.sample(n=min(len(df_disc), 10000), random_state=42)
    xlim = df_disc[discount_col].quantile(0.99)
    ylim = df_disc['TotalAmount'].quantile(0.99)

    # Binned aggregation: order count, total revenue, average order value by discount bin
    bins = [0, 1, 5, 10, 15, 20, 25, 30, 40, 50, 100]
    df_disc['disc_bin'] = pd.cut(df_disc[discount_col], bins=bins, include_lowest=True)
    agg = df_disc.groupby('disc_bin').agg(
        OrderCount=('OrderID', 'nunique'),
        TotalRevenue=('TotalAmount', 'sum'),
        AvgOrderValue=('TotalAmount', 'mean')
    ).reset_index()

    # Plotting
    plt.figure(figsize=(14,6))
    # Scatter + regression
    plt.subplot(1,2,1)
    sns.scatterplot(data=sample, x=discount_col, y='TotalAmount', alpha=0.2, s=10)
    sns.regplot(data=sample, x=discount_col, y='TotalAmount', scatter=False, color='red', line_kws={'linewidth':1.5})
    plt.xlim(0, max(5, xlim))  # keep some room for small discounts
    plt.ylim(0, ylim)
    plt.xlabel('Discount (%)')
    plt.ylabel('TotalAmount ($)')
    plt.title('Discount % vs Order Value (scatter + linear fit)')

    # Binned: Order count (bars) and Avg order value (line)
    plt.subplot(1,2,2)
    ax = sns.barplot(data=agg, x=agg['disc_bin'].astype(str), y='OrderCount', color='skyblue')
    ax2 = ax.twinx()
    sns.lineplot(data=agg, x=agg['disc_bin'].astype(str), y='AvgOrderValue', color='orange', marker='o', ax=ax2)
    ax.set_xticklabels([str(x.get_text()) for x in ax.get_xticklabels()], rotation=45)
    ax.set_xlabel('Discount Bin (%)')
    ax.set_ylabel('Order Count')
    ax2.set_ylabel('Avg Order Value ($)')
    plt.title('Order Count and Avg Order Value by Discount Bin')

    plt.tight_layout()
    plt.show()

    # Print aggregated table for quick inspection
    print(agg.to_string(index=False))
```

**INSIGHTS** : The analysis shows a weak negative correlation (-0.108) between discount percentage and TotalAmount, indicating that higher discounts do not necessarily lead to higher sales value. In fact, orders with lower discounts (0–1%) account for the highest sales volume and revenue, with 40,246 orders and a total revenue of $39.76 million, along with the highest average order value ($987.91). As discount levels increase, both order count and average order value consistently decline, suggesting that deeper discounts may attract fewer or lower-value purchases rather than driving volume growth. This pattern implies that customers are more responsive to product value and pricing than to aggressive discounting, and that moderate or minimal discounts may be more effective in sustaining higher sales volume and revenue.

# 14. Which products should be promoted or discounted more? 

```python
# Product promotion / discount recommendation visualization
prod_metrics = amazondata.groupby(product_col).agg(
    OrderCount=('OrderID', 'nunique'),
    TotalRevenue=('TotalAmount', 'sum'),
    AvgOrderValue=('TotalAmount', 'mean'),
    AvgDiscount=('Discount', 'mean'),
    AvgUnitPrice=('UnitPrice', 'mean'),
    Cancelled=('OrderStatus', lambda s: (s == 'Cancelled').sum()),
    Returned=('OrderStatus', lambda s: (s == 'Returned').sum())
).reset_index()

prod_metrics['CancelRate'] = prod_metrics['Cancelled'] / prod_metrics['OrderCount']
prod_metrics['ReturnRate'] = prod_metrics['Returned'] / prod_metrics['OrderCount']

# thresholds (medians)
ord_med = prod_metrics['OrderCount'].median()
rev_med = prod_metrics['TotalRevenue'].median()
disc_med = prod_metrics['AvgDiscount'].median()
price_med = prod_metrics['AvgUnitPrice'].median()

# candidate selection rules
promote_candidates = prod_metrics[
    (prod_metrics['OrderCount'] >= ord_med) &
    (prod_metrics['TotalRevenue'] >= rev_med) &
    (prod_metrics['AvgDiscount'] <= disc_med)
].sort_values('TotalRevenue', ascending=False).head(10)

discount_candidates = prod_metrics[
    (prod_metrics['OrderCount'] <= ord_med) &
    (prod_metrics['TotalRevenue'] >= rev_med) &
    (prod_metrics['AvgDiscount'] <= disc_med) &
    (prod_metrics['AvgUnitPrice'] >= price_med)
].sort_values('AvgUnitPrice', ascending=False).head(10)

# Scatter plot: popularity vs revenue, color by avg discount, size by avg order value
plt.figure(figsize=(14,9))
sc = plt.scatter(
    prod_metrics['OrderCount'], prod_metrics['TotalRevenue'],
    c=prod_metrics['AvgDiscount'], cmap='coolwarm', s=(prod_metrics['AvgOrderValue'] / prod_metrics['AvgOrderValue'].max()) * 400 + 20,
    alpha=0.75, edgecolors='k', linewidth=0.3
)
plt.colorbar(sc, label='Avg Discount (%)')
plt.xscale('linear')
plt.yscale('linear')
plt.xlabel('Order Count')
plt.ylabel('Total Revenue ($)')
plt.title('Product Promotion / Discount Candidates (size ~ AvgOrderValue, color ~ AvgDiscount)')

# median lines to form quadrants
plt.axvline(ord_med, color='gray', linestyle='--')
plt.axhline(rev_med, color='gray', linestyle='--')

# annotate top promote and discount candidates
for i, row in promote_candidates.head(5).iterrows():
    plt.annotate(row[product_col], (row['OrderCount'], row['TotalRevenue']), xytext=(8, -6), textcoords='offset points', fontsize=9, color='green')

for i, row in discount_candidates.head(5).iterrows():
    plt.annotate(row[product_col], (row['OrderCount'], row['TotalRevenue']), xytext=(8, 6), textcoords='offset points', fontsize=9, color='red')

plt.tight_layout()
plt.show()

# Print recommendations
print("Top Promote Candidates (popular & high revenue, low discount):")
print(promote_candidates[[product_col, 'OrderCount', 'TotalRevenue', 'AvgDiscount', 'AvgOrderValue']].to_string(index=False))

print("\nTop Discount Candidates (high revenue but lower popularity; consider discounts to increase volume):")
print(discount_candidates[[product_col, 'OrderCount', 'TotalRevenue', 'AvgUnitPrice', 'AvgDiscount', 'AvgOrderValue']].to_string(index=False))
```

**INSIGHTS** : The data suggests a clear distinction between products that should be promoted versus those that would benefit from additional discounting.

Products such as Electric Kettle, Dress Shirt, Water Bottle, Gaming Mouse, and Cookware Set emerge as strong promotion candidates because they combine high order volumes (≈2,000+ orders), high total revenue (≈$1.85M–$1.91M), and relatively low average discounts (~7%). Their consistently high average order values indicate strong customer demand even without aggressive price reductions, making them ideal for increased visibility through marketing campaigns, featured placements, or bundling strategies rather than deeper discounts.

On the other hand, items like Smartwatch, Smartphone Case, Kids Toy Car, Jeans, Laptop Sleeve, and HDMI Cable 2m generate high revenue but slightly lower popularity, suggesting untapped demand. These products are better suited as discount candidates, where targeted or time-limited discounts could stimulate higher purchase volume without significantly eroding revenue. Overall, this strategy balances profitability by promoting already high-performing products while using selective discounts to unlock growth in underperforming yet valuable items.

# 15. Is there a correlation between UnitPrice and Discount?

```python
# Correlation & visualization: UnitPrice vs Discount
if 'UnitPrice' in amazondata.columns and 'Discount' in amazondata.columns:
    df_ud = amazondata[['UnitPrice', 'Discount']].dropna().copy()
    # ensure Discount is in percent for readability
    if df_ud['Discount'].max() <= 1:
        df_ud['DiscountPct'] = df_ud['Discount'] * 100
    else:
        df_ud['DiscountPct'] = df_ud['Discount'].astype(float)

    # correlation metrics
    pearson_corr = df_ud['UnitPrice'].corr(df_ud['DiscountPct'])
    spearman_corr = df_ud['UnitPrice'].corr(df_ud['DiscountPct'], method='spearman')

    # scatter with regression (sample for speed)
    sample = df_ud.sample(n=min(len(df_ud), 15000), random_state=42)
    plt.figure(figsize=(14,6))

    plt.subplot(1,2,1)
    sns.scatterplot(data=sample, x='UnitPrice', y='DiscountPct', alpha=0.18, s=14)
    sns.regplot(data=sample, x='UnitPrice', y='DiscountPct', scatter=False, color='red', line_kws={'linewidth':1.5})
    plt.xlim(0, df_ud['UnitPrice'].quantile(0.99))
    plt.ylim(0, df_ud['DiscountPct'].quantile(0.995))
    plt.xlabel('UnitPrice ($)')
    plt.ylabel('Discount (%)')
    plt.title(f'UnitPrice vs Discount (%)   Pearson={pearson_corr:.3f}, Spearman={spearman_corr:.3f}')

    # Binned analysis (qcut)
    df_ud['price_bin'] = pd.qcut(df_ud['UnitPrice'], q=10, duplicates='drop')
    agg = df_ud.groupby('price_bin').agg(AvgDiscount=('DiscountPct', 'mean'), MedianDiscount=('DiscountPct','median'), Count=('DiscountPct','count')).reset_index()

    plt.subplot(1,2,2)
    ax = sns.barplot(data=agg, x=agg.index, y='AvgDiscount', color='skyblue')
    ax2 = ax.twinx()
    sns.lineplot(data=agg, x=agg.index, y='Count', marker='o', color='orange', ax=ax2)
    ax.set_xticklabels([f'{str(r)}' for r in agg['price_bin']], rotation=45, ha='right')
    ax.set_xlabel('UnitPrice bins (quantiles)')
    ax.set_ylabel('Avg Discount (%)')
    ax2.set_ylabel('Order Count')
    plt.title('Avg Discount & Order Count by UnitPrice Quantile Bins')
    plt.tight_layout()
    plt.show()

    print(f'Pearson correlation: {pearson_corr:.3f}')
    print(f'Spearman correlation: {spearman_corr:.3f}')
else:
    print("Columns 'UnitPrice' and/or 'Discount' not found in amazondata.")
```

**INSIGHTS** : The correlation analysis shows virtually no relationship between Unit Price and Discount in the dataset. Both the Pearson (0.006) and Spearman (0.006) correlation coefficients are close to zero, indicating that higher-priced products are not systematically associated with higher or lower discounts. This suggests that discounting strategies are applied independently of product price, likely driven by factors such as promotions, inventory management, or marketing campaigns rather than unit price levels.
