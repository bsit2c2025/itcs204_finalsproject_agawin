**Output Display**: Conclude the plotting commands with plt.show() to render the completed figure in the Jupyter Notebook cell output.

### Visualization on the 1st Matrix : Piano Keyboard Frequency Mapping (Hz)


```python
import numpy as np

import matplotlib.pyplot as plt

# Visualize frequency_matrix as a heatmap
fig, ax = plt.subplots(figsize=(6, 3.5))
cax = ax.imshow(frequency_matrix, cmap='magma', aspect='auto')

col_labels = ['C', 'C#', 'D', 'D#']
row_labels = ['C4–D#4', 'E4–G4', 'G#4–B4']

ax.set_xticks(np.arange(len(col_labels)))
ax.set_xticklabels(col_labels)
ax.set_yticks(np.arange(len(row_labels)))
ax.set_yticklabels(row_labels)

# Annotate each cell with frequency value
for i in range(frequency_matrix.shape[0]):
    for j in range(frequency_matrix.shape[1]):
        ax.text(j, i, f"{frequency_matrix[i, j]:.2f}", ha='center', va='center', color='white', fontsize=8)

ax.set_title("Piano Keyboard Frequency Mapping (Hz)")
ax.set_xlabel("Note")
ax.set_ylabel("Range")
fig.colorbar(cax, ax=ax, label='Hz')
plt.tight_layout()
plt.show()
```

**INSIGHTS** : The heatmap clearly illustrates the systematic increase in frequency as notes ascend both chromatically and across pitch ranges, reflecting the mathematical structure of Western music tuning.

**Consistent Frequency Progression:** <br>
Moving from left to right (C → C# → D → D#), frequencies increase steadily, demonstrating the equal-tempered scale where each semitone represents a uniform frequency ratio.

**Octave-Based Growth:** <br>
From the top row (C4–D#4) to the bottom row (G#4–B4), frequencies rise significantly, highlighting how pitch doubles approximately every octave—an essential principle in music acoustics.

**Visual Clarity of Pitch Hierarchy:** <br>
Darker colors represent lower frequencies while brighter tones indicate higher frequencies, making it easy to visually distinguish low, mid, and high pitch regions on the keyboard.

**Data Structure Relevance:** <br>
The matrix format efficiently models a 2D relationship between notes and pitch ranges, which is ideal for algorithmic applications such as sound synthesis, pitch detection, or music recommendation systems.

Overall, this visualization demonstrates how structured numerical data can accurately represent musical concepts, reinforcing the importance of matrices and data organization when analyzing or generating music programmatically.

### Visualization on the 2nd Matrix : 2nd Matrix : Chord Construction (Notes as Pitch Classes)

```python
# Visualize chord_matrix as a heatmap with pitch-class annotations
pitch_names = ['C','C#','D','D#','E','F','F#','G','G#','A','A#','B']
chord_labels = ['C Major', 'D Minor', 'E Minor', 'F Major']

fig2, ax2 = plt.subplots(figsize=(6, 3))
cax2 = ax2.imshow(chord_matrix, cmap='viridis', aspect='auto')

ax2.set_xticks(np.arange(chord_matrix.shape[1]))
ax2.set_xticklabels([pitch_names[i] for i in range(chord_matrix.shape[1])])
ax2.set_yticks(np.arange(chord_matrix.shape[0]))
ax2.set_yticklabels(chord_labels)

for r in range(chord_matrix.shape[0]):
    for c in range(chord_matrix.shape[1]):
        val = int(chord_matrix[r, c])
        note = pitch_names[val % 12]
        text_color = 'white' if chord_matrix[r, c] > chord_matrix.mean() else 'black'
        ax2.text(c, r, f"{val}\n{note}", ha='center', va='center', color=text_color, fontsize=9)

ax2.set_title("Chord Construction (Pitch Classes)")
ax2.set_xlabel("Pitch Class")
ax2.set_ylabel("Chord")
fig2.colorbar(cax2, ax=ax2, label='Pitch Class Value')
plt.tight_layout()
plt.show()
```

**INSIGHTS** : The visualization effectively demonstrates how chords are built using pitch classes and interval structures, making abstract music theory concepts concrete through numerical representation.

**Interval-Based Chord Formation:** <br>
Each row represents a chord constructed from its root using fixed pitch-class intervals. For example, C Major (0–4–7) clearly follows the major triad formula, while D Minor (2–5–9) reflects the minor triad pattern. This highlights how chord quality is determined by interval spacing rather than absolute note names.

**Repetition Indicates Octave Equivalence:** <br>
The repeated pitch class values (e.g., C appearing as both 0 and again at the end of a row) emphasize the concept of octave equivalence, where notes separated by octaves share the same pitch class.

**Color Intensity Reveals Harmonic Distance:** <br>
Lighter colors correspond to higher pitch class values, visually distinguishing chord tones such as the leading tone B (11) in E Minor. This helps identify harmonic tension and resolution tendencies within chords.

**Structured Matrix Representation:** <br>
Representing chords in a matrix format allows for efficient computational operations such as pattern matching, transposition, and chord comparison, which are essential in music analysis algorithms and digital audio applications.

Overall, the visualization shows how data structures like matrices can accurately encode harmonic relationships, bridging music theory and algorithmic processing in a clear and intuitive way.
