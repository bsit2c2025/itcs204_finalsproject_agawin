**Pandas for Data Handling** <br> 

**Pandas** is the essential library for managing and analyzing structured data in Python. Its core data structure is the **DataFrame**, which is a two-dimensional, labeled table (like a spreadsheet or SQL table) that can hold different types of data (numeric, string, boolean, etc.) in columns.



Before you can visualize complex data, you need to load it, clean it, and structure it. Pandas makes this process efficient. It allows you to read files (CSV, Excel), select specific columns, filter rows, and prepare the data specifically for statistical plotting. For example, you might load a dataset of house prices and use Pandas to select only the 'size' and 'price' columns, ready for plotting.

**Key Pandas Operations** <br>

* **Data Loading:** Reading data from external sources (e.g., pd.read_csv('data.csv')). <br>
* **Inspection:** Examining the first few rows (df.head()) and checking data types (df.info()). <br>
* **Preparation:** Selecting, filtering, and grouping data for analysis. <br>

**Seaborn for Statistical Visualization**



Seaborn is a powerful data visualization library built on top of Matplotlib. While Matplotlib provides the low-level plotting tools, Seaborn offers a high-level interface for drawing informative statistical graphics. It is especially effective when working directly with Pandas DataFrames.



Seaborn simplifies the creation of sophisticated plots like box plots, violin plots, and relationship plots, which are often used in data analysis to understand distributions and relationships between variables. By passing an entire Pandas DataFrame to a Seaborn function, you can specify columns for the X and Y axes, and Seaborn automatically handles the data mapping and provides visually appealing defaults.

### 1. Student performance varies significantly across class levels

```python
import matplotlib.pyplot as plt

# Create a visualization showing performance variation across class levels
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Box plot for Performance Score by Class Level
sns.boxplot(data=musicdata, x='Class_Level', y='Performance_Score', ax=axes[0, 0], palette='Set2')
axes[0, 0].set_title('Performance Score by Class Level', fontsize=12, fontweight='bold')
axes[0, 0].set_ylabel('Performance Score')

# Box plot for Skill Development by Class Level
sns.boxplot(data=musicdata, x='Class_Level', y='Skill_Development', ax=axes[0, 1], palette='Set2')
axes[0, 1].set_title('Skill Development by Class Level', fontsize=12, fontweight='bold')
axes[0, 1].set_ylabel('Skill Development Rating')

# Box plot for Accuracy by Class Level
sns.boxplot(data=musicdata, x='Class_Level', y='Accuracy', ax=axes[1, 0], palette='Set2')
axes[1, 0].set_title('Accuracy by Class Level', fontsize=12, fontweight='bold')
axes[1, 0].set_ylabel('Accuracy (%)')

# Box plot for Engagement Score by Class Level
sns.boxplot(data=musicdata, x='Class_Level', y='Engagement_Score', ax=axes[1, 1], palette='Set2')
axes[1, 1].set_title('Engagement Score by Class Level', fontsize=12, fontweight='bold')
axes[1, 1].set_ylabel('Engagement Score')

plt.tight_layout()
plt.show()

# Print summary statistics
print("\nSummary Statistics by Class Level:\n")
print(musicdata.groupby('Class_Level')[['Performance_Score', 'Skill_Development', 'Accuracy', 'Engagement_Score']].describe().round(2))
```

**INSIGHTS** : The visualization clearly demonstrates that student performance differs meaningfully across Beginner, Intermediate, and Advanced class levels. The Advanced group shows a slightly higher median performance score compared to the other levels, indicating stronger overall academic outcomes as students progress. However, it also exhibits a wider spread, suggesting variability in mastery even among advanced learners.

The Beginner group displays a lower median performance with noticeable dispersion, reflecting mixed foundational understanding and the natural adjustment phase at early stages of learning. In contrast, Intermediate students show a more balanced distribution, with performance scores clustering closer to the median, implying greater consistency as core concepts become more established.

Taken together, these patterns indicate that progression through class levels is associated with improvements in performance, but also highlight that higher levels introduce greater differentiation among students. This insight supports the need for level-specific instructional strategies, where beginners benefit from foundational reinforcement, intermediates from consolidation, and advanced students from more personalized or challenge-based learning approaches.

### 2. Rhythm and pitch accuracy show a strong relationship with overall performance

```python
# Create a visualization showing relationship between Rhythm, Pitch Accuracy, and Performance Score
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Scatter plot: Rhythm vs Performance Score with regression line
sns.regplot(data=musicdata, x='Rhythm', y='Performance_Score', ax=axes[0, 0], scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
axes[0, 0].set_title('Rhythm vs Performance Score', fontsize=12, fontweight='bold')
axes[0, 0].set_xlabel('Rhythm')
axes[0, 0].set_ylabel('Performance Score')

# Scatter plot: Pitch Accuracy vs Performance Score with regression line
sns.regplot(data=musicdata, x='Pitch_Accuracy', y='Performance_Score', ax=axes[0, 1], scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
axes[0, 1].set_title('Pitch Accuracy vs Performance Score', fontsize=12, fontweight='bold')
axes[0, 1].set_xlabel('Pitch Accuracy')
axes[0, 1].set_ylabel('Performance Score')

# Heatmap: Correlation between Rhythm, Pitch Accuracy, and Performance Score
correlation_data = musicdata[['Rhythm', 'Pitch_Accuracy', 'Performance_Score', 'Accuracy', 'Skill_Development']].corr()
sns.heatmap(correlation_data, annot=True, fmt='.2f', cmap='coolwarm', ax=axes[1, 0], cbar_kws={'label': 'Correlation'})
axes[1, 0].set_title('Correlation Matrix: Performance Metrics', fontsize=12, fontweight='bold')

# Joint scatter plot: Rhythm and Pitch Accuracy colored by Performance Score
scatter = axes[1, 1].scatter(musicdata['Rhythm'], musicdata['Pitch_Accuracy'], c=musicdata['Performance_Score'], cmap='viridis', s=100, alpha=0.6)
axes[1, 1].set_title('Rhythm & Pitch Accuracy (colored by Performance Score)', fontsize=12, fontweight='bold')
axes[1, 1].set_xlabel('Rhythm')
axes[1, 1].set_ylabel('Pitch Accuracy')
plt.colorbar(scatter, ax=axes[1, 1], label='Performance Score')

plt.tight_layout()
plt.show()

# Print correlation coefficients
print("\nCorrelation with Performance Score:\n")
print(musicdata[['Rhythm', 'Pitch_Accuracy', 'Accuracy', 'Skill_Development', 'Performance_Score']].corr()['Performance_Score'].sort_values(ascending=False))
```

**INSIGHTS** : Overall performance appears only weakly influenced by rhythm and pitch accuracy, suggesting that high technical scores alone do not guarantee higher performance outcomes.

* **Rhythm vs Performance:** The slight upward trend and near-zero correlation (~0.03) indicate that rhythm consistency has minimal direct impact on performance score. Performers with both low and high rhythm scores achieve similar performance levels.

* **Pitch Accuracy vs Performance:** The regression line slopes downward and the correlation is mildly negative (~-0.19), implying that higher pitch accuracy does not necessarily translate into better performance, and may even coincide with slightly lower performance scores in this dataset—possibly due to over-focusing on technical precision at the expense of expressiveness.

* **Combined View (Rhythm & Pitch):** The colored scatter shows high and low performance scores spread across all rhythm–pitch combinations. This dispersion reinforces that performance is multidimensional, likely driven by qualitative factors not captured here (e.g., expression, confidence, interpretation).

* **Correlation Matrix Takeaway:** No strong correlations exist between performance score and any single metric. This suggests performance evaluation is holistic, and improvements may require balanced development rather than optimizing one technical skill.

**Key takeaway:** <br>
Performance excellence is not dominated by rhythm or pitch accuracy alone—artistry, interpretation, and overall musicality likely play a larger role than isolated technical metrics.

### 3. Engagement score is influenced by lesson type (Theory vs Practical)

```python
# Visualization: Engagement Score by Lesson Type
fig, ax = plt.subplots(figsize=(8, 6))
sns.boxplot(data=musicdata, x='Lesson_Type', y='Engagement_Score', palette='Set2', ax=ax)
sns.stripplot(data=musicdata, x='Lesson_Type', y='Engagement_Score', color='k', alpha=0.5, jitter=True, ax=ax)
ax.set_title('Engagement Score by Lesson Type (Theory vs Practical)', fontsize=12, fontweight='bold')
ax.set_xlabel('Lesson Type')
ax.set_ylabel('Engagement Score')
plt.tight_layout()
plt.show()

# Summary statistics
print(musicdata.groupby('Lesson_Type')['Engagement_Score'].agg(['count', 'mean', 'median', 'std']).round(2))
```

**INSIGHTS** : The visualization suggests that practical lessons tend to generate slightly higher and more consistent engagement than theory lessons. While both lesson types cover a wide engagement range (roughly 1–10), the median engagement score for practical lessons is higher, indicating that students are generally more engaged during hands-on activities. In contrast, theory lessons show greater variability, with more extreme low and high scores, implying that theoretical content resonates strongly with some students but disengages others. Overall, this pattern highlights the value of practical, applied learning in sustaining student engagement.

___

### 4. Instrument type affects tempo and accuracy metrics


```python
# Visualization: Instrument type affects Tempo and Accuracy
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.boxplot(data=musicdata, x='Instrument_Type', y='Tempo', palette='Set3', ax=axes[0])
sns.stripplot(data=musicdata, x='Instrument_Type', y='Tempo', color='k', alpha=0.5, jitter=True, ax=axes[0])
axes[0].set_title('Tempo by Instrument Type')
axes[0].set_xlabel('Instrument Type')
axes[0].set_ylabel('Tempo')
axes[0].tick_params(axis='x', rotation=45)

sns.boxplot(data=musicdata, x='Instrument_Type', y='Accuracy', palette='Set2', ax=axes[1])
sns.stripplot(data=musicdata, x='Instrument_Type', y='Accuracy', color='k', alpha=0.5, jitter=True, ax=axes[1])
axes[1].set_title('Accuracy by Instrument Type')
axes[1].set_xlabel('Instrument Type')
axes[1].set_ylabel('Accuracy')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# Summary statistics by instrument
print(musicdata.groupby('Instrument_Type')[['Tempo', 'Accuracy']].agg(['count', 'mean', 'median', 'std']).round(2))
```

**INSIGHTS** : The visualization indicates clear performance differences across instrument types. Violin players tend to perform at the highest tempos, showing both a higher median tempo and a wider spread, which suggests technical demands that encourage faster playing but with greater variability. Guitar and piano tempos are slightly lower and more clustered, indicating more consistency in pacing.

In terms of accuracy, piano players achieve the highest median accuracy, suggesting greater precision and control, while violin accuracy is slightly lower and more variable, likely reflecting the instrument’s sensitivity to technique. Guitar accuracy shows the widest spread, implying that performance quality varies more by player skill level. Overall, the charts suggest a trade-off: instruments associated with higher tempos (like violin) show more variability, while those emphasizing control (like piano) demonstrate more consistent accuracy.
